#02 조건문

## 02-1 if문
| 구문          | 의미                |
|----------------|---------------------|
| if ~	           |만약 조건을 만족하면 ~ 작업 수행|
| if ~ else ~ |만약 조건을 만족하면 ~ 작업 수행하고, 그렇지 않으면 ~ 작업을 수행|
| if ~ elif ~ else    |만약 조건 1을 만족하면 작업 1 수행, 조건 2를 만족하면 작업 2 수행... 그렇지 않으면 else 작업 수행|
#엘리프


```python
# 사용자가 입력한 정수 값이, 음수 양수를 처리하는 조건문을 만들어보자
x = int(input('정수를 입력하세요 : '))


if x > 0 :
  print('입력된 수는 양수입니다.')
elif x == 0:
  print('입력된 값은 0입니다.')
else:
  print('입력된 수는 음수입니다.')


'''
x = 0 : x에 0이 들어간다.
x == 0 : x는 0이다.

엘리프 함수는 조건이 여러 가지인 경우
if else 함수는 조건이 2가지인 경우
if 함수는 그 외에 경우는 처리를 안 함
'''
```

몇 번 횟수 반복할지 지정하는 것

```python
range(10)

#0부터 10까지의 구간이다.
```

```python
for i in range(10): # 0 ~ 9 / i를, 10번째 구간까지(0~9까지) 반복함.
  print(i)

'''
숫자로 반복시키기 위해서는 range
i 가 변수임.. i말고 a 써도 되고, 변수에 뭘 넣어도 상관은 없음.
range 의 10가지 값을 변수에서 반복한다
0등부터 9등까지의 번호표를 발급하기 위해
숫자형 반복은 for ~ range
range함수는 항상 0부터 시작하기 때문에 +1
'''

for i in range(1, 11): # 1 ~ 10 (range는 원하는 숫자로 합하려면 +1을 해야함)
  print(i)
#1부터 10까지만 반복할 것이다.. range 함수를 통해 반복시킨다. 반복시킬 수 있는 횟수를 지정.
```

##03-2 while문

조건이 만족하는 동안에 반복문을 실행

(조건만 만족하면 계속 반복반복.
조건에서 탈출할 브레이크가 없으면 무한반복.
탈출 못하면 메모리를 꺼야함 크롬브라우저를 꺼야함. 주의!!

for문은 딱 반복할 구간만 설정하는 것)

와일 문은 로직 안에서만 반복

```python
# 1부터 10까지 합계를 구하라
total = 0
i = 1

#(if문이랑 비슷하게, 그러나 10보다 커질 수 있게 탈출구를 만들어야 함)
while i <= 10:
  total = total + i
  print('i의 값 : %d => 합계 : %d' % (i, total))

# 탈출코드 -> i값을 증가시켜서 조건 비교 할 수 있도록 구현(탈출할 수 있도록)
  i = i + 1

print('1부터 10까지의 합은 %d입니다.' % total)
# = print('1부터 10까지의 합은 %d입니다.' % (total))
```

```python
# for문을 이용해서 1부터 10까지의 합계를 구하라

total = 0
i = 1

for i in range(1,11): #10까지 반복하는 것임 range함수는 항상 0부터 시작하기 때문에 +1
  total = total + i

  if(i == 5):
    break

  print('1부터 10까지의 합은 %d입니다.' % total)
```

### break문 반복문 탈출하기

```python
# 사용자가 입력한 단어를 출력하고,  stop 이라는 단어 입력 시 프로그램 종료
while True:    #와일 트루는 조건 없이 반복문 무한 반복 (게임 자동화 등에 사용)

  word = input('단어를 입력하세요: ')

  if word == 'stop':
    print('프로그램을 종료합니다')
    break

  print(word)
```

### for문을 통한 딕셔너리 사용

```python
# 딕셔너리를 이용해서 영어 단어 퀴즈 프로그램
# stop 단어를 입력하면 퀴즈 프로그램이 종료될 수 있는 조건 추가
words = {'꽃':'flower','나비':'butterfly', '학교':'school', '자동차':'car'}

print('*** 영어 단어 퀴즈 게임 시작 ***')

for key in words: #딕셔너리의 for문은 변수에 key 값을 넘겨줌)
  input_word = input('%s에 해당되는 영어 단어를 입력하세요' % key)

  if input_word == 'stop':
    print('프로그램을 종료합니다')
    break #break를 안 넣으면 안 끝남. STOP을 IF문으로 먼저 빼야, 인공지능이 왔다갔다 안 함.stop을 if문에서 가장 위로 올렸음.
  elif input_word == words[key]:
    print('정답입니다')
  else:
    print('오답입니다')

print('*** 게임 종료 ***')

'''
딕셔너리는 key:value의 사전 형태로 저장이 되기 때문에, 내가 이름으로 정보를 찾기 수월하다.
그래서 인덱스는 모든 정보를 일차원적으로 순서대로 좌아악 나열하기 때문에, 정보를 나열하기는 쉽지만
정보를 찾기에는 딕셔너리가 수월하다.
딕셔너리는 무조건 key를 줘야지만 조건문을 사용할 수 있으니, key를 주는 조건문을 만들어라.
값은 앞에 적은 키를 통해서만 추출 가능.
'''
```

```python
import random #랜덤으로 만들어주는 함수

# 영어 단어 퀴즈가 무한 반복으로 진행되고, 사용자가 stop 키워드를 입력해야지만 퀴즈가 종료되게 설계 (while문)
#while - true 로 반복문으로 바로 진입 가능

#딕셔너리 키들만 따로 저장 (퀴즈 문제 만들기)
words = {'꽃':'flower','나비':'butterfly', '학교':'school', '자동차':'car'}

words_keys = list(words.keys())

print('*** 영어 단어 퀴즈 게임 시작 ***')

while True:
  key = random.choice(words_keys) #랜덤.초이스하면 알아서 키 중에 랜덤으로 뽑음
  input_word = input('%s에 해당되는 영어 단어를 입력하세요' % key)

  if input_word == 'stop':
    print('프로그램을 종료합니다')
    break #break를 안 넣으면 안 끝남. STOP을 IF문으로 먼저 빼야, 인공지능이 왔다갔다 안 함.
  elif input_word == words[key]:
    print('정답입니다')
  else:
    print('오답입니다')


print('*** 게임 종료 ***')
```
