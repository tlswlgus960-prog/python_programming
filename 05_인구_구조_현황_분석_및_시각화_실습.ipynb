###행정코드 삭제하기(원본 건드리지 말고,)

```python
population_df['행정구역'] = population_df['행정구역'].str.split('(').str.get(0).str.strip() # 이제 이거를 df에 담아라!
```

### 총인구수 데이터 특수기호 (,콤마) 지우기 - `replace()` 메소드를 사용하여 삭제.

콤마를 없앤다음에 형변환을 해야 오류가 없음

결측치 등등 모든 오류를 다 처리한 후에, 형변환을 하는 게 순서이다. 오류가 없도록.

```python
# 특수기호(,) 삭제하기 - replace (바꾸기) 메소드
#replace(찾을 문자, 바꿀 문자, regex=True, inplace=True)
population_df.replace(',','', regex=True, inplace=True) #','콤마를 -> ''아무것도 아닌걸로 바꿔달라 -  replace(찾을 문자, 바꿀 문자) / 정규식.정규표현식. 레귤러 익스프레션의 약자. regex=True 특수기호를 나타내는 정규식. 특수기호를 뽑거나, 특수기호만 없애거나. 정규표현식을 적용해서 바꿀거다, 라는 뜻임. / inplace=True 바꾼 거 바로 적용해달라! 명령
population_df
```

### 결측치 (naN = not a number) 처리하기 - 공백이 아니고, 숫자 처리가 안 된 것임.

```python
#먼저, 결측치가 몇 건이나 있는지 확인부터 해야 함
population_df.isna().sum() #이즈엔에이. 트루 폴스로 진행 ->sum 다 더해. sum은 true만 더한다!
```

### 강원특별자치도와 강원도 인구수 데이터 병합하기

```python
population_df[['강원특별자치도', '강원도']] #2023년도부터 바뀐 거 확인.[]한개 확인, [[]]2개,여러개 확인. -> 강원도를 강원특별자치도로 병합시켜야겠다. 를 확인할 것.
```

```python
# naN 데이터(아직 정의되지 않은 값, 연산이 불가능) 처리 -> 강원도 컬럼 데이터를 강원특별자치도 컬럼으로 통합
# 처리 순서: naN 데이터 (값을 채우거나, 값을 지우거나 해야 함) -> 형변환 -> 0으로 채우기 -> 강원도 컬럼 데이터를 강원특별자치도로 통합(더하는)

population_df['강원특별자치도'] = population_df['강원도'].astype(float).fillna(0) + population_df['강원특별자치도'].astype(float).fillna(0)  #float로 형변환(실수가 없어서, 정수로 해도되는거아닌가? -> 실수가> 정수보다 크니까, 먼저 가장 큰 범위의 실수로 변환시키고, 그 다음 정수로 변환시키는 게 좋다.) /fillna 필엔에이. 채워주는 함수 - 0으로 채워줘 / dropna() - ()로 없애줘.
population_df['강원특별자치도']
#서로 float로, 0으로 채운 후, 두개를 합치고, 합친 거를 강원특별자치도로 넣어라!
```

###연도별 대한민국 전체 인구 변환

17개 시 도의 인구를 합산하여 대한민군 전체 인구가 연도별로 어떻게 변화했는지 확인

```python
# 연도별 총 인구수 (17개 시도를 합친)를 집계 (새로이 만들어야 함. 모든 지역의 인구수 총합.)
# 1. 17개 시도 컬럼명 리스트
region_col = pop_df.columns.drop('연도') #연도를 빼보렴 . 지금 쓸 데이터에 연도는 안 필요해서.
region_col

# 2. 연도별, 지역별 인구를 모두 합하여 전국 인구 파생 변수를 생성
pop_df = pop_df.assign(전국인구=pop_df[region_col].sum(axis=1)) # assign: 맡기다,파견하다 _ 새 열을 만드는. 파생변수. / axis=1: 엑세스. 컬럼단위(y축끼리)로 연산하고싶어요 = 0, 연도단위로 (row단위로 x축끼리.) 연산하고싶어요 = 1
pop_df
#assign()은 기존 DataFrame의 열을 이용하여 새로운 열(파생 변수)을 추가하거나 계산된 값을 새로운 열로 만들어 반환하는 메서드
```

### 2015년 대비 2024년 지역별 인구 증강 분석하기


```python
# 2015년 데이터와 2024년 데이터를 준비해야함(y축 데이터는 냅두고, x축 데이터만 15년, 24년도꺼만 남기면 두 개를 비교하기 쉽다.)

#1. 2015년 데이터 추출하기 ()튜플로 만들어서, 수정 불가능하게 보안.
pop_2015 = (pop_df.query("연도 == '2015년_총인구수'").drop(columns=['연도', '전국인구']).T.reset_index()) #쿼리로 이것만 찾아라! 함수. / drop(columns==['연도', '전국인구'] 연도와 전국인구를 지워라. / T.reset-index() : T트랜스포 다시 바꿔. 표에 붙여놓을 인덱스 형태로. 리셋 초기화시켜. 인덱스 권한 삭제시킬게. 너네 컬럼으로 와! 인덱스는 표에 안 들어가니까, 컬럼으로 변환시킴. 그 옆에 0~10으로 인덱스 번호가 생김.

#컬럼명 정리
pop_2015.columns = ['지역', '2015년 인구'] #컬럼명 변경
pop_2015
```

```python
#2. 2024년 데이터 추출하기 ()튜플로 만들어서, 수정 불가능하게 보안.
pop_2024 = (pop_df.query("연도 == '2024년_총인구수'").drop(columns=['연도', '전국인구']).T.reset_index()) #쿼리로 이것만 찾아라! 함수. / drop(columns==['연도', '전국인구'] 연도와 전국인구를 지워라. / T.reset-index() : T트랜스포 다시 바꿔. 표에 붙여놓을 인덱스 형태로. 리셋 초기화시켜. 인덱스 권한 삭제시킬게. 너네 컬럼으로 와! 인덱스는 표에 안 들어가니까, 컬럼으로 변환시킴. 그 옆에 0~10으로 인덱스 번호가 생김.

#컬럼명 정리
pop_2024.columns = ['지역', '2024년 인구'] #컬럼명 변경
pop_2024
```

```python
# 3. 2015, 2024년 데이터 프레임 결합
pop_change = pd.merge(pop_2015, pop_2024, on='지역') #merge 멀지 두 개를 합치는 거. 기준이 있어야 함. on=지역<기준> on이 기준.
pop_change
```

```python
#4. 인구증감을 계산하여, 파생변수 만들기 (인구증감(+,-) 추가하기)
pop_change = pop_change.assign(인구증감 = pop_change['2024년 인구'] - pop_change['2015년 인구'])
pop_change
```

```python
# 5. 인구 증감 순으로 정렬 (데이터 기반으로 정렬 기준 세우기)
pop_change = pop_change.sort_values(by='인구증감',  ascending=False)  #ascending=False 로 해줘야 인구 증가한 순서대로 내려온다. - asc가 오름차순, desc가 내림차순./asc를 false로 하면 내림차순으로.
pop_change
```

```python
# 6. 인구증감 시각화

#인구증감 시각화 코드

# 시각화

fig = px.bar(
    pop_change,
    x='인구증감',
    y='지역',
    orientation='h',
    title='2015년 대비 2024년 지역별 인구 증감',
    text='인구증감',
    color='인구증감',
    color_continuous_scale='RdBu' #색깔의 스케일을 조정한다. 플러스가 강하면 파랑색, 마이너스가 강하면 빨간색으로. re = red/ bu=blue
)

fig.update_layout(
    xaxis_title='인구 증감(명)',
    yaxis_title='지역',
    yaxis={'categoryorder': 'total ascending'},
    template='plotly_white',
    height=700,
    coloraxis_showscale=False
)

fig.update_traces(
    texttemplate='%{text:+,}',
    textposition='outside'
)

fig.add_vline(
    x=0,
    line_width=1,
    line_dash='dash'
)

fig.show()
```

### 전년 대비 인구 증감 분석

`diff()` 메서드를 사용해서 전년 대비 전국 인구가 얼마나 증가, 감소했는지 분석하기

```python
pop_df = pop_df.assign(전년대비증감=pop_df['전국인구'].diff()) #diff: 각각 전년대비 컬럼에 맞게 정리해줌. different 함수. row끼리 빼는 거. 상위 row에서 하위 row를 순서대로 빼준다.(바로 위 줄에서 바로 아래 줄을 빼는 함수.)
pop_df
```

```python
# 시각화하기_전년 대비 증감 시각화 (2015년부터 있던 데이터에 추가한 것.)

# 전년대비 증감 시각화
# 시각화

fig = px.bar(
    pop_df,
    x='연도',
    y='전년대비증감',
    title='전년 대비 전국 인구 증감',
    text='전년대비증감'
)

fig.update_layout(
    xaxis_title='연도',
    yaxis_title='인구 증감(명)',
    template='plotly_white'
)

fig.update_traces(
    texttemplate='%{text:+,}',
    textposition='outside'
)

fig.add_hline(
    y=0,
    line_dash='dash'
)

fig.show()
```

### 전국 인구의 증감율 분석
`pct_change()` 메서드를 이용하여 전년 대비 인구 증가율(%)을 분석하기

몇 명이 증가했나 잘 안 함. 몇 퍼센트 증가했나르 주로 사용함.

```python
pop_df = pop_df.assign(전년대비증감률=pop_df['전국인구'].pct_change()*100) #pct_change()*100: 퍼센트 비율을 내는 함수. 무조건 *100 해야함. 비율을 찾아야하니까. 바로 위에 diff와 같은 원리인데, 비율 계산임. 바로 상위 줄에서 하위 줄 빠진 거를 몇 % 줄었는지 비율로 확인하는 것.
pop_df
```

```python
# 시각화. 전년 대비 증감율.

#전년대비 증감률 시각화
# 시각화

fig = px.line(
    pop_df,
    x='연도',
    y='전년대비증감률',
    markers=True,
    title='전년 대비 전국 인구 증감률'
)

fig.update_layout(
    xaxis_title='연도',
    yaxis_title='증감률(%)',
    template='plotly_white'
)

fig.add_hline(
    y=0,
    line_dash='dash'
)

fig.show()
```

### 전국의 인구 이동 평균 분석
`rolling().mean()` 메서드를 이용하여 단기 변화를 줄이고 장기적인 인구 추세를 확인

이동평균=추세를 분석하는 거.

```python
pop_df = pop_df.assign(이동평균3년=pop_df['전국인구'].rolling(3).mean())  #rolling 3년치
pop_df
```

```python
#이동평균 시각화

# 시각화

fig = px.line(
    pop_df,
    x='연도',
    y=['전국인구', '이동평균3년'],
    markers=True,
    title='전국 인구와 3년 이동평균'
)

fig.update_layout(
    xaxis_title='연도',
    yaxis_title='인구수',
    template='plotly_white'
)

fig.show()
```

이동평균은 "추세(Trend)를 보기 위해" 사용

이번 데이터는 연도별 변동이 크지 않아 이동평균과 원본 데이터의 차이가 크지 않지만, 이동평균은 주식 가격, 일별 방문자 수, 월별 매출처럼 변동이 큰 데이터를 분석할 때 더욱 효과적으로 활용

* 데이터는 해마다 조금씩 증가하거나 감소하는 등 단기적인 변동이 발생

* 이러한 변동 때문에 전체적인 흐름을 파악하기 어려울 수 있음

* 일정 기간의 평균을 계산하여 전체적인 추세를 부드럽게 확인

* 주식 차트의 5일선, 20일선, 60일선이 모두 이동평균
