```python
# 라인스타일 추가

import matplotlib.pyplot as plt

plt.title('plotting') #타이틀 속성 추가

#1개 이상의 차트를 같이 그릴 경우, label 속성 추가, 색상 변경, 마커 추가 (.으로 하면 됨 '.'), 라인스타일 추가
plt.plot([100, 200, 300, 400, 500],[10, 20, 30, 40, 50], linestyle='--', marker='^', color='r', label='asc') #label 속성 추가(asc 상승하는 / desc 하강하는)
plt.plot([100, 200, 300, 400, 500],[50, 40, 30, 20, 10], ls='--', marker='.', color='b', label='desc') #ls=라인스타일

#지피티는 앞으로 마커, 컬러, 라벨, 라인스타일 등을 이렇게 명시함. 'r^--' 이게 라인스타일인지 , 컬러인지 등을 파악해라.

#범례 속성 추가
plt.legend() #범례 추가
plt.show()
```

##03 서울 기온 데이터 시각화

### 최고 기온 시각화

```python
import csv
import matplotlib.pyplot as plt
#위 두 줄 먼저 작성 후 실행 먼저. 그 후 아래 작성.

# 1) 파일을 연다.
f = open('/content/seoul.csv', 'r', encoding='euc-kr')

# 2) 파일을 읽는다.
data = csv.reader(f, delimiter=',')  #f (=는 담긴거고, f에 오픈이 담겨있으니까) 리더로 읽으려고 적음. f로 열었는데, delimiter=','로 읽어라. 라는 명령. 콤마로 구분되어있으니, 읽어봐라. 라는 뜻.
print(data)

# 3) 헤더 저장 (헤더 한 줄 넘기기)
header = next(data) # next() 한줄 읽고 커서(마우스 선택)를 다음으로 넘김_읽고 있는 파일의 위치. next 하면 다음 줄로 커서가 넘어가기 때문에, 데이터가 2번째 줄부터 읽는다.

# 4) 최고기온을 담을 리스트 변수를 설정
result = []

# 5) 결측치를 제외한 최고 기온 데이터만 result 리스트에 담기(결측치를 제외하고 모든 데이터를 담을 것.)
for row in data :
  if row[-1] != '': #결측치가 아니면, 결과값에 담는다.
    result.append(float(row[-1]))  #리저트 리스트에 어팬드 붙인다. 플롯 row형도 변형시킨다. 오류 안 나게.

#6) 파일 닫기 (파일을 열었으니, 닫으라는 명령 해야 함)
f.close()

#7) 차트 그리기
plt.title('seoul')
plt.plot(result, 'r')
plt.show()

# 기간을 전체로 두면 의미 없는 결과값이 나올 수도 있다. 내가 의미 있는 자료를 어떻게 설정할지 고민.
```
